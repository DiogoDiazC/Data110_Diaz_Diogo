```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
fig = plt.figure()
ax = plt.axes()
```

```python
x = np.linspace(0, 5, 10)
ax.plot(x, np.sin(x));
```

```python
plt.plot(x, np.sin(x));
```

```python
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x));
```

```python
plt.plot(x, np.sin(x - 0), color='blue')
plt.plot(x, np.sin(x - 1), color='g')
plt.plot(x, np.sin(x - 2), color='0.75')
plt.plot(x, np.sin(x - 3), color='#FFDD44')
plt.plot(x, np.sin(x - 4), color=(1.0,0.2,0.3))
plt.plot(x, np.sin(x - 5), color='chartreuse');
```

```python
plt.plot(x, x + 4, linestyle='dashed')
plt.plot(x, x + 5, linestyle='dashdot')
plt.plot(x, x + 7, linestyle='solid')
plt.plot(x, x + 9, linestyle='dotted');
```

```python
plt.plot(x, x + 4, '-g')   # solid red
plt.plot(x, x + 5, '--c')  # dashed blue
plt.plot(x, x + 7, '-.k')  # dashdot green
plt.plot(x, x + 9, ':r');  # dotted orange
```

```python
plt.plot(x, np.sin(x))

plt.xlim(-4, 12)
plt.ylim(-3.3, 3.3);
```

```python
plt.plot(x, np.sin(x))

plt.xlim(5, 0)
plt.ylim(2.8, -2.8);
```

```python
plt.plot(x, np.sin(x))
plt.axis('tight');
```

```python
plt.plot(x, np.sin(x))
plt.axis('equal');
```

```python
plt.plot(x, np.sin(x))
plt.title("A Sine Curve")
plt.xlabel("x")
plt.ylabel("sin(x)");
```

```python
plt.plot(x, np.sin(x), '-g', label='sin(x)')
plt.plot(x, np.cos(x), ':b', label='cos(x)')
plt.axis('equal')

plt.legend();
```

This gives graphs a much bigger picture of how certain values change, but ti doesn't give away any major points to the study. That's why it's better to utilize both graphing types.

```python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
plt.style.use("ggplot")
```

```python
x = [1, 2, 3, 4]
y = [3, 7, 14, 5]

plt.bar(x, y, label='Week 1')
plt.legend()
plt.show()
```

```python
x = ['a','b','c','d']
y = [3, 7, 14, 5]
plt.bar(x, y, label='Week 1')
plt.legend()
plt.show()
```

```python
x = ['p1', 'p2', 'p3', 'p4']
y = [3, 7, 14, 5]
z = [11, 9, 15, 20]
plt.bar(x, y, label='Week 1')
plt.bar(x, z, label='Week 2', bottom=y)  # Stacks z on top of y
plt.legend()
plt.show()
```

```python
import numpy as np


x = ['Games', 'Movies', 'Books']
y = [7, 10, 4]  # Data for Week 1
z = [9, 8, 5]  # Data for Week 2

x_indexes = np.arange(len(x))
print(x_indexes)
plt.bar(x_indexes - 0.125, y, width=0.25, label='Week 1')

plt.bar(x_indexes + 0.125, z, width=0.25, label='Week 2')

plt.xticks(ticks=x_indexes, labels=x)

plt.legend()

plt.show()
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/Reben80/Data110-22016/refs/heads/main/dataset/happiness_2017.csv')
df.head()
df.info()

plt.figure(figsize=(25, 25))
plt.barh(df['Country'], df['HappinessScore'])
plt.show()
```

```python
df_sorted = df.sort_values(by='HappinessScore', ascending=True)
plt.figure(figsize=(10, 15))
plt.barh(df_sorted['Country'], df_sorted['HappinessScore'])
plt.xlabel('Happiness Score')
plt.ylabel('Country')
plt.title('Happiness Score by Country')

ax = plt.gca()
ax.tick_params(axis='y', labelsize=8)

plt.show()
```

```python
df_top_10 = df_sorted.head(10)
plt.figure(figsize=(10, 5))

plt.barh(df_top_10['Country'], df_top_10['HappinessScore'])

plt.xlabel('Happiness Score')
plt.ylabel('Country')
plt.title('Happiness Score by Country')

ax = plt.gca()
ax.tick_params(axis='y', labelsize=8)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.barh(df_sorted.tail(10)['Country'], df_sorted.tail(10)['HappinessScore'])

plt.xlabel('Happiness Score')
plt.ylabel('Country')
plt.title('Top 10 Countries by Happiness Score')

ax = plt.gca()
ax.tick_params(axis='y', labelsize=8)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data = {'Movie': ['Star Wars', 'Jumanji', 'Pitch Perfect 3', 'Greatest Showman', 'Ferdinand'],
        'Earnings': [71.57, 36.17, 19.93, 8.81, 7.32]}

df = pd.DataFrame(data)

df_sorted = df.sort_values(by='Earnings', ascending=True)

plt.style.use('fivethirtyeight')
plt.figure(figsize=(10, 6))
plt.barh(df_sorted['Movie'], df_sorted['Earnings'], color='skyblue',zorder=2)

plt.xlabel('Earnings (in millions)')
plt.ylabel('Movie Title')
plt.title('Movie Box Office Earnings')

ax = plt.gca()
ax.tick_params(axis='y', labelsize=8)
plt.grid(axis='y', linestyle='-', color='gray', zorder=0)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.barh(titles, amounts, color='skyblue', zorder=3)

plt.title('Movie Box Office Earnings')
plt.xlabel('Amount (in millions)')
plt.ylabel('Movie Title')

ax = plt.gca()

ax.set_xlim(0, 80)

plt.xticks([0, 20, 40, 60])

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)

plt.grid(axis='x', linestyle='-', color='#d3d3d3', zorder=0)

plt.tight_layout()
plt.show()
```

```python
data = {'Movie': ['Star Wars', 'Jumanji', 'Pitch Perfect 3', 'Greatest Showman', 'Ferdinand'],
        'Earnings': [71.57, 36.17, 19.93, 8.81, 7.32]}
dm = pd.DataFrame(data)

dm_sorted = dm.sort_values(by='Earnings', ascending=True)

plt.style.use('fivethirtyeight')
plt.figure(figsize=(10, 6))
plt.barh(dm_sorted['Movie'], dm_sorted['Earnings'], color='skyblue')

plt.xlabel('Earnings (in millions)')
plt.ylabel('Movie Title')
plt.title('Movie Box Office Earnings')

plt.xticks([0, 20, 40, 60])


plt.grid(axis='y', linestyle='-', color='#d3d3d3', zorder=0)

ax = plt.gca()
ax.tick_params(axis='y', labelsize=8)

plt.tight_layout()
plt.show()
```
